# Tạo enities

```python
import pandas as pd
from tqdm import tqdm


def build_entities(filtered_meta, doc_to_chunks_map):
    """
    Tạo danh sách entity đại diện cho từng văn bản pháp luật.

    Parameters
    ----------
    filtered_meta : pandas.DataFrame
        DataFrame chứa metadata của các văn bản pháp luật.

    doc_to_chunks_map : dict
        Dictionary ánh xạ từ số hiệu văn bản sang danh sách chunk_id.
        Ví dụ:
        {
            "12/2024/NĐ-CP": ["chunk_001", "chunk_002"]
        }

    Returns
    -------
    list
        Danh sách các entity theo định dạng phù hợp để import vào
        hệ thống Knowledge Graph hoặc LightRAG.
    """

    # Hiển thị tiêu đề của bước xử lý
    print("\n" + "=" * 60)
    print("BƯỚC 3: Tạo Entities (LegalDocument)")
    print("=" * 60)

    # Danh sách lưu các entity được tạo
    entities = []

    # Duyệt qua từng dòng metadata
    # tqdm được dùng để hiển thị thanh tiến trình
    for _, row in tqdm(
        filtered_meta.iterrows(),
        total=len(filtered_meta),
        desc="Đang tạo entities"
    ):
        # Lấy loại văn bản, ví dụ: Luật, Nghị định, Thông tư
        loai_vb = row.get("Loại Văn Bản")

        # Bỏ qua nếu văn bản không có thông tin loại văn bản
        if pd.isna(loai_vb):
            continue

        # Dùng số hiệu văn bản làm định danh của document
        doc_id = row.get("Số Hiệu")

        # Bỏ qua nếu không có số hiệu
        if pd.isna(doc_id):
            continue

        # Chuẩn hóa số hiệu về chuỗi và loại bỏ khoảng trắng thừa
        doc_id = str(doc_id).strip()

        # Bỏ qua giá trị rỗng hoặc chuỗi "nan"
        if not doc_id or doc_id.lower() == "nan":
            continue

        # Tạo phần mô tả tổng hợp cho entity từ metadata
        description_parts = []

        # Thêm loại văn bản
        description_parts.append(str(loai_vb).strip())

        # Thêm số hiệu văn bản
        if pd.notna(row.get("Số Hiệu")):
            description_parts.append(str(row["Số Hiệu"]).strip())

        # Thêm chủ đề hoặc lĩnh vực của văn bản
        if pd.notna(row.get("Chủ Đề")):
            description_parts.append(str(row["Chủ Đề"]).strip())

        # Thêm cơ quan hoặc nơi ban hành văn bản
        if pd.notna(row.get("Nơi Ban Hành")):
            description_parts.append(
                f"Được ban hành bởi: {str(row['Nơi Ban Hành']).strip()}"
            )

        # Thêm thông tin hiệu lực của văn bản
        if pd.notna(row.get("Hiệu lực")):
            description_parts.append(str(row["Hiệu lực"]).strip())

        # Ghép các thành phần thành một đoạn mô tả hoàn chỉnh
        description = ". ".join(description_parts)

        # Lấy danh sách các chunk thuộc văn bản hiện tại
        chunk_ids = doc_to_chunks_map.get(doc_id, [])

        # Nếu văn bản có chunk, nối các chunk_id bằng ký hiệu <SEP>
        # Nếu không có chunk, dùng số hiệu văn bản làm source_id dự phòng
        source_id = "<SEP>".join(map(str, chunk_ids)) if chunk_ids else doc_id

        # Lấy đường dẫn đến file nguồn
        # Dùng giá trị rỗng nếu cột file_path không tồn tại hoặc bị thiếu
        file_path = row.get("file_path", "")
        file_path = "" if pd.isna(file_path) else str(file_path)

        # Tạo entity LegalDocument
        entity = {
            # Tên entity là số hiệu văn bản
            "entity_name": doc_id,

            # Loại entity dùng để phân biệt với Article, Agency, Topic...
            "entity_type": "LegalDocument",

            # Mô tả tổng hợp từ metadata của văn bản
            "description": description,

            # Danh sách các chunk chứa nội dung của văn bản
            "source_id": source_id,

            # Đường dẫn đến file dữ liệu nguồn
            "file_path": file_path,
        }

        # Thêm entity vào danh sách kết quả
        entities.append(entity)

    # Thông báo tổng số entity đã tạo
    print(f"  → Tạo {len(entities)} entities")

    # Trả về danh sách entity
    return entities
```

# Build Relationship

```python
# Bảng ánh xạ tên mối quan hệ trong dữ liệu gốc
# sang từ khóa mô tả và trọng số của cạnh trong Knowledge Graph.
REL_MAP = {
    # Văn bản hiện tại hướng dẫn hoặc quy định chi tiết
    # cho một văn bản pháp luật khác.
    "Văn bản hướng dẫn": {
        "kw": "hướng dẫn, quy định chi tiết",
        "w": 0.95
    },

    # Văn bản liên quan được sử dụng làm căn cứ pháp lý
    # để ban hành văn bản hiện tại.
    "Văn bản căn cứ": {
        "kw": "căn cứ, cơ sở pháp lý",
        "w": 1.0
    },

    # Văn bản hiện tại là đối tượng được một văn bản khác
    # hướng dẫn hoặc quy định chi tiết.
    "Văn bản được hướng dẫn": {
        "kw": "được hướng dẫn, được quy định chi tiết",
        "w": 0.95
    },

    # Văn bản hiện tại quy định một phần nội dung của
    # văn bản liên quan không còn hiệu lực.
    "Văn bản quy định hết hiệu lực một phần": {
        "kw": "bãi bỏ, hết hiệu lực một phần",
        "w": 0.85
    },

    # Một phần nội dung của văn bản hiện tại đã hết hiệu lực
    # do tác động của văn bản liên quan.
    "Văn bản hết hiệu lực một phần": {
        "kw": "hết hiệu lực một phần",
        "w": 0.8
    },

    # Văn bản hiện tại được đưa vào một văn bản hợp nhất.
    "Văn bản được hợp nhất": {
        "kw": "được hợp nhất",
        "w": 0.75
    },

    # Văn bản hiện tại sửa đổi hoặc bổ sung
    # nội dung của văn bản liên quan.
    "Văn bản sửa đổi, bổ sung": {
        "kw": "sửa đổi, bổ sung",
        "w": 0.9
    },

    # Văn bản hiện tại có dẫn chiếu hoặc viện dẫn
    # đến văn bản liên quan.
    "Văn bản dẫn chiếu": {
        "kw": "dẫn chiếu, tham chiếu, viện dẫn",
        "w": 0.7
    },

    # Văn bản hiện tại bị văn bản liên quan
    # sửa đổi hoặc bổ sung.
    "Văn bản bị sửa đổi, bổ sung": {
        "kw": "bị sửa đổi, bổ sung",
        "w": 0.9
    },

    # Văn bản hiện tại đã hết hiệu lực hoặc bị bãi bỏ
    # bởi văn bản liên quan.
    "Văn bản hết hiệu lực": {
        "kw": "hết hiệu lực, bị bãi bỏ",
        "w": 0.85
    },

    # Văn bản hiện tại thay thế văn bản liên quan.
    "Văn bản thay thế": {
        "kw": "thay thế",
        "w": 0.9
    },

    # Văn bản hiện tại vừa được hợp nhất,
    # vừa có nội dung sửa đổi hoặc bổ sung.
    "Văn bản được hợp nhất (sửa đổi)": {
        "kw": "được hợp nhất, sửa đổi",
        "w": 0.8
    },
}


def build_relationships(df_rels, doc_to_chunks_map):
    """
    Tạo danh sách các mối quan hệ giữa các văn bản pháp luật.

    Parameters
    ----------
    df_rels : pandas.DataFrame
        DataFrame chứa thông tin quan hệ giữa các văn bản.

        Các cột được sử dụng:
        - "mối quan hệ": Loại quan hệ giữa hai văn bản.
        - "số ký hiệu": Số hiệu văn bản nguồn.
        - "số ký hiệu văn bản liên quan": Số hiệu văn bản đích.

    doc_to_chunks_map : dict
        Dictionary ánh xạ số hiệu văn bản với danh sách chunk_id.

        Ví dụ:
        {
            "12/2024/NĐ-CP": [
                "chunk_001",
                "chunk_002"
            ]
        }

    Returns
    -------
    list
        Danh sách relationship phù hợp để đưa vào
        Knowledge Graph hoặc LightRAG.
    """

    # Hiển thị tiêu đề của bước xử lý
    print("\n" + "=" * 60)
    print("BƯỚC 4: Tạo Relationships từ bảng rels")
    print("=" * 60)

    # Danh sách lưu các relationship hợp lệ
    relationships = []

    # Đếm số dòng bị bỏ qua do thiếu dữ liệu,
    # sai loại quan hệ hoặc tạo cạnh tự tham chiếu
    skipped = 0

    # Duyệt qua từng dòng trong bảng quan hệ
    # tqdm dùng để hiển thị thanh tiến trình
    for _, row in tqdm(
        df_rels.iterrows(),
        total=len(df_rels),
        desc="Đang tạo relationships"
    ):
        # Lấy tên loại quan hệ bằng tiếng Việt
        rel_type_vi = str(row.get("mối quan hệ", "")).strip()

        # Tra cứu thông tin quan hệ trong REL_MAP
        rel_info = REL_MAP.get(rel_type_vi)

        # Bỏ qua nếu loại quan hệ chưa được định nghĩa
        if not rel_info:
            skipped += 1
            continue

        # Lấy số hiệu của văn bản nguồn
        doc_id = str(row.get("số ký hiệu", "")).strip()

        # Lấy số hiệu của văn bản đích hoặc văn bản liên quan
        tgt_id = str(
            row.get("số ký hiệu văn bản liên quan", "")
        ).strip()

        # Kiểm tra tính hợp lệ của hai đầu mối quan hệ
        invalid_relationship = (
            not doc_id
            or not tgt_id
            or doc_id.lower() == "nan"
            or tgt_id.lower() == "nan"
            or doc_id == tgt_id
        )

        # Bỏ qua nếu thiếu số hiệu hoặc văn bản tự liên kết với chính nó
        if invalid_relationship:
            skipped += 1
            continue

        # Lấy các chunk thuộc văn bản nguồn
        chunk_ids = doc_to_chunks_map.get(doc_id, [])

        # Nếu văn bản có nhiều chunk thì nối các chunk_id bằng <SEP>
        # Nếu không có chunk thì dùng số hiệu văn bản làm source_id dự phòng
        source_id = (
            "<SEP>".join(map(str, chunk_ids))
            if chunk_ids
            else doc_id
        )

        # Lấy động từ chính từ chuỗi từ khóa
        # Ví dụ:
        # "hướng dẫn, quy định chi tiết" -> "hướng dẫn"
        primary_relation = rel_info["kw"].split(",")[0].strip()

        # Tạo một cạnh có hướng:
        # doc_id ──relationship──> tgt_id
        relationship = {
            # ID của entity nguồn
            "src_id": doc_id,

            # ID của entity đích
            "tgt_id": tgt_id,

            # Câu mô tả ngắn về quan hệ giữa hai văn bản
            "description": (
                f"{doc_id} {primary_relation} {tgt_id}"
            ),

            # Các từ khóa ngữ nghĩa dùng cho quá trình truy xuất
            "keywords": rel_info["kw"],

            # Trọng số biểu thị mức độ quan trọng
            # hoặc độ mạnh của mối quan hệ
            "weight": rel_info["w"],

            # Các chunk chứa thông tin của văn bản nguồn
            "source_id": source_id,

            # Thông tin dùng để truy ngược về nguồn
            # Hiện tại dùng số hiệu văn bản nguồn
            "file_path": doc_id,
        }

        # Thêm relationship vào danh sách kết quả
        relationships.append(relationship)

    # In kết quả tổng hợp
    print(
        f"  → Tạo {len(relationships)} relationships "
        f"(bỏ {skipped} dòng)"
    )

    # Trả về danh sách các mối quan hệ đã tạo
    return relationships
```

# Build Chunks

```python
import re
import pandas as pd

from tqdm import tqdm
from pyvi import ViTokenizer
from langchain_experimental.text_splitter import SemanticChunker
from langchain_huggingface import HuggingFaceEmbeddings


# ============================================================
# 1. KHỞI TẠO MÔ HÌNH EMBEDDING
# ============================================================

# Chạy mô hình embedding bằng CPU.
# Có thể đổi thành {"device": "cuda"} nếu máy có GPU hỗ trợ CUDA.
model_kwargs = {
    "device": "cpu"
}

# Chuẩn hóa vector embedding về độ dài 1.
# Việc này phù hợp khi sử dụng cosine similarity.
encode_kwargs = {
    "normalize_embeddings": True
}

# Mô hình embedding đã được fine-tune cho dữ liệu pháp luật tiếng Việt.
embeddings_Model = HuggingFaceEmbeddings(
    model_name="MinhNguyet3001/embedding-supper-legal",
    model_kwargs=model_kwargs,
    encode_kwargs=encode_kwargs
)


# ============================================================
# 2. KHỞI TẠO SEMANTIC CHUNKER
# ============================================================

# SemanticChunker chia văn bản dựa trên mức độ thay đổi ngữ nghĩa
# giữa các câu thay vì chỉ cắt theo số lượng ký tự cố định.
chunker = SemanticChunker(
    # Mô hình dùng để tạo embedding cho từng câu.
    embeddings=embeddings_Model,

    # Số câu lân cận được đưa vào khi tính ngữ cảnh embedding.
    # Buffer càng lớn thì việc xác định điểm ngắt càng chú ý
    # đến nội dung trước và sau câu hiện tại.
    buffer_size=5,

    # Dùng độ biến thiên gradient của khoảng cách embedding
    # để tìm điểm chuyển chủ đề.
    breakpoint_threshold_type="gradient",

    # Ngưỡng xác định điểm ngắt semantic.
    # Giá trị càng cao thì chunk thường càng lớn và ít bị chia nhỏ.
    breakpoint_threshold_amount=0.75,

    # Tách câu tại dấu chấm, hỏi hoặc cảm thán.
    # Phần (?!\d+\.\s) giúp hạn chế tách nhầm tại danh sách đánh số
    # như "1. Nội dung", "2. Nội dung".
    sentence_split_regex=r"(?<=[.?!])\s+(?!\d+\.\s)",

    # Kích thước tối thiểu của một semantic chunk.
    min_chunk_size=512
)


# ============================================================
# 3. HÀM ĐẾM TOKEN TIẾNG VIỆT
# ============================================================

def count_tokens(text):
    """
    Ước lượng số token của văn bản tiếng Việt.

    ViTokenizer được sử dụng để tách từ ghép tiếng Việt.
    Sau đó số token được tính bằng số phần tử sau khi split.

    Ví dụ:
        "quyền sử dụng đất"
        có thể được tách thành:
        "quyền sử_dụng đất"

    Parameters
    ----------
    text : str
        Nội dung cần đếm token.

    Returns
    -------
    int
        Số token ước lượng.
    """
    return len(ViTokenizer.tokenize(text).split())


# ============================================================
# 4. CHIA NỘI DUNG MỘT ĐIỀU LUẬT THEO NGỮ NGHĨA
# ============================================================

def split_article_body(full_content):
    """
    Chia nội dung của một Điều thành các semantic sub-chunk.

    Quy tắc:
    - Nội dung rỗng: trả về một chuỗi rỗng.
    - Nội dung không quá 1.000 token: giữ nguyên một chunk.
    - Nội dung trên 1.000 token: sử dụng SemanticChunker.
    - Nếu SemanticChunker gặp lỗi: trả lại nguyên văn bản.

    Parameters
    ----------
    full_content : str
        Toàn bộ phần nội dung nằm trong một Điều luật.

    Returns
    -------
    list[str]
        Danh sách các phần nội dung đã được chia.
    """

    # Loại bỏ khoảng trắng ở đầu và cuối nội dung.
    clean_content = full_content.strip()

    # Nếu Điều không có nội dung thì trả về chunk rỗng.
    if not clean_content:
        return [""]

    # Ước lượng độ dài nội dung bằng tokenizer tiếng Việt.
    token_len = count_tokens(clean_content)

    # Những Điều ngắn được giữ nguyên để tránh chia nhỏ
    # và làm mất tính toàn vẹn của ngữ cảnh pháp lý.
    if token_len <= 1000:
        return [clean_content]

    try:
        # Chỉ những Điều dài trên 1.000 token mới được chia semantic.
        semantic_parts = [
            part.strip()
            for part in chunker.split_text(clean_content)
            if part and part.strip()
        ]

    except Exception:
        # Nếu mô hình embedding hoặc SemanticChunker gặp lỗi,
        # giữ nguyên Điều để không làm mất dữ liệu.
        return [clean_content]

    # Nếu SemanticChunker không tạo được kết quả hợp lệ,
    # sử dụng lại nội dung ban đầu.
    return semantic_parts or [clean_content]


# ============================================================
# 5. PARSE MỘT VĂN BẢN THÀNH CÁC LEGAL CHUNK
# ============================================================

def parse_to_chunks(content, so_ky_hieu):
    """
    Phân tích nội dung một văn bản pháp luật thành các chunk theo Điều.

    Cấu trúc pháp lý được nhận diện gồm:

        Văn bản
        └── Chương
            └── Mục
                └── Tiểu mục
                    └── Điều
                        └── Semantic sub-chunk

    Mỗi Điều được tạo thành một chunk. Nếu Điều quá dài,
    nội dung tiếp tục được chia thành các semantic sub-chunk.

    Parameters
    ----------
    content : str
        Nội dung văn bản pháp luật đã được chuyển sang plain text.

    so_ky_hieu : str
        Số ký hiệu của văn bản, ví dụ:
        "12/2024/NĐ-CP".

    Returns
    -------
    list[dict]
        Danh sách chunk của văn bản.
    """

    # Không xử lý nếu nội dung bị thiếu hoặc có giá trị NaN.
    if not content or pd.isna(content):
        return []

    # Trường hợp dữ liệu đầu vào còn ở dạng HTML có thể sử dụng:
    #
    # soup = BeautifulSoup(content, "html.parser")
    # text = soup.get_text(separator="\n")
    #
    # Trong pipeline hiện tại, Content đã là plain text
    # nên chỉ cần tách trực tiếp theo dòng.

    # Tách nội dung thành từng dòng và loại bỏ dòng trống.
    lines = [
        line.strip()
        for line in content.split("\n")
        if line.strip()
    ]

    # Lưu Chương hiện tại mà vòng lặp đang xử lý.
    current_chuong = ""

    # Lưu Mục hiện tại.
    current_muc = ""

    # Lưu Tiểu mục hiện tại.
    current_tieu_muc = ""

    # Lưu tiêu đề Điều hiện tại, ví dụ:
    # "Điều 15. Điều kiện cấp giấy phép".
    current_dieu = ""

    # Lưu các dòng nội dung thuộc Điều hiện tại.
    current_content = []

    # Danh sách kết quả chứa toàn bộ chunk của văn bản.
    chunks = []

    # Nhận diện tiêu đề Chương, ví dụ:
    # "Chương I", "CHƯƠNG 2", "Chương IV:"
    chuong_pattern = re.compile(
        r"^(Chương|CHƯƠNG)\s+([IVXLCDM]+|\d+)(\.|:)?"
    )

    # Nhận diện tiêu đề Mục, ví dụ:
    # "Mục 1", "MỤC II:"
    muc_pattern = re.compile(
        r"^(Mục|MỤC)\s+([IVXLCDM]+|\d+)(\.|:)?"
    )

    # Nhận diện tiêu đề Tiểu mục.
    tieu_muc_pattern = re.compile(
        r"^(Tiểu mục|TIỂU MỤC)\s+([IVXLCDM]+|\d+)(\.|:)?"
    )

    # Nhận diện Điều luật, ví dụ:
    # "Điều 1.", "Điều 25."
    dieu_pattern = re.compile(r"^Điều\s+(\d+)\.")


    def save_current_dieu():
        """
        Lưu Điều hiện tại vào danh sách chunks.

        Hàm nội bộ này được gọi khi:
        - Phát hiện một Điều mới.
        - Đã duyệt hết toàn bộ văn bản.
        """

        nonlocal current_dieu, current_content

        # Không tạo chunk nếu chưa phát hiện Điều nào.
        if not current_dieu:
            return

        # Ghép toàn bộ các dòng thuộc Điều thành một nội dung hoàn chỉnh.
        full_content = "\n".join(current_content).strip()

        # Prefix giúp bảo toàn cấu trúc phân cấp pháp lý trong chunk:
        # [Số ký hiệu] [Chương] [Mục] [Tiểu mục]
        prefix = f"[{so_ky_hieu}]"

        if current_chuong:
            prefix += f" [{current_chuong.strip()}]"

        if current_muc:
            prefix += f" [{current_muc.strip()}]"

        if current_tieu_muc:
            prefix += f" [{current_tieu_muc.strip()}]"

        # Lấy số Điều từ tiêu đề Điều hiện tại.
        # Regex này hỗ trợ cả trường hợp Điều 10a, Điều 10b.
        dieu_match = re.search(
            r"Điều\s+(\d+[a-z]?)",
            current_dieu,
            flags=re.IGNORECASE
        )

        # Nếu không nhận diện được thì dùng "0" làm giá trị dự phòng.
        dieu_num = dieu_match.group(1) if dieu_match else "0"

        # Loại bỏ ký tự chữ để tạo thứ tự số.
        # Ví dụ:
        # "10a" -> 10
        chunk_order_index = int(
            re.sub(r"[a-z]", "", dieu_num, flags=re.IGNORECASE)
        )

        # Nếu Điều dài, chia tiếp thành semantic sub-chunks.
        semantic_parts = split_article_body(full_content)

        # ID cơ sở của chunk được tạo từ:
        # số ký hiệu văn bản + số Điều.
        base_source_id = f"chunk_{so_ky_hieu}_{dieu_num}"

        # Tạo từng chunk hoặc semantic sub-chunk.
        for part_index, part_content in enumerate(
            semantic_parts,
            start=1
        ):
            # Nếu Điều chỉ có một phần thì giữ source_id gốc.
            source_id = base_source_id

            # Nếu Điều được chia thành nhiều phần,
            # thêm hậu tố _s1, _s2, _s3...
            if len(semantic_parts) > 1:
                source_id = f"{base_source_id}_s{part_index}"

            # Nội dung chunk bao gồm:
            # cấu trúc phân cấp + tiêu đề Điều + nội dung Điều.
            chunk_text = (
                f"{prefix} {current_dieu}\n{part_content}"
            ).strip()

            # Thêm chunk vào danh sách kết quả.
            chunks.append(
                {
                    # Nội dung được dùng để embedding và truy xuất.
                    "content": chunk_text,

                    # Định danh duy nhất của chunk.
                    "source_id": source_id,

                    # Thứ tự Điều trong văn bản.
                    "chunk_order_index": chunk_order_index,

                    # Đường dẫn logic để truy ngược về văn bản nguồn.
                    "file_path": so_ky_hieu,
                }
            )

    # Duyệt lần lượt từng dòng của văn bản.
    for line in lines:
        # --------------------------------------------------------
        # Kiểm tra tiêu đề Chương
        # --------------------------------------------------------
        chuong_match = chuong_pattern.match(line)

        if chuong_match:
            # Cập nhật Chương hiện tại.
            current_chuong = line

            # Khi sang Chương mới, reset Mục và Tiểu mục cũ.
            current_muc = ""
            current_tieu_muc = ""
            continue

        # --------------------------------------------------------
        # Kiểm tra tiêu đề Mục
        # --------------------------------------------------------
        muc_match = muc_pattern.match(line)

        if muc_match:
            # Cập nhật Mục hiện tại.
            current_muc = line

            # Khi sang Mục mới, reset Tiểu mục cũ.
            current_tieu_muc = ""
            continue

        # --------------------------------------------------------
        # Kiểm tra tiêu đề Tiểu mục
        # --------------------------------------------------------
        tieu_muc_match = tieu_muc_pattern.match(line)

        if tieu_muc_match:
            current_tieu_muc = line
            continue

        # --------------------------------------------------------
        # Kiểm tra tiêu đề Điều
        # --------------------------------------------------------
        dieu_match = dieu_pattern.match(line)

        if dieu_match:
            # Khi bắt gặp Điều mới, lưu Điều trước đó.
            save_current_dieu()

            # Thiết lập Điều hiện tại là dòng vừa nhận diện.
            current_dieu = line

            # Reset nội dung để bắt đầu thu thập nội dung của Điều mới.
            current_content = []

        else:
            # Những dòng không phải tiêu đề cấu trúc sẽ được xem
            # là phần nội dung của Điều hiện tại.
            if current_dieu:
                current_content.append(line)

    # Sau khi vòng lặp kết thúc, cần lưu Điều cuối cùng
    # vì không còn Điều mới để kích hoạt save_current_dieu().
    save_current_dieu()

    return chunks


# ============================================================
# 6. TẠO CHUNK CHO TOÀN BỘ TẬP VĂN BẢN
# ============================================================

def build_chunks(df_content):
    """
    Parse toàn bộ văn bản trong DataFrame thành legal chunks.

    Mỗi văn bản được chia theo cấu trúc Điều luật. Các Điều dài
    tiếp tục được chia thành semantic sub-chunk.

    Parameters
    ----------
    df_content : pandas.DataFrame
        DataFrame chứa tối thiểu hai cột:

        - "Số Ký Hiệu": Số hiệu của văn bản.
        - "Content": Nội dung plain text của văn bản.

    Returns
    -------
    tuple
        all_chunks : list[dict]
            Toàn bộ chunk được tạo từ các văn bản.

        doc_to_chunks_map : dict
            Ánh xạ từ số hiệu văn bản đến danh sách source_id.

            Ví dụ:
            {
                "12/2024/NĐ-CP": [
                    "chunk_12/2024/NĐ-CP_1",
                    "chunk_12/2024/NĐ-CP_2"
                ]
            }
    """

    # Hiển thị tiêu đề bước xử lý.
    print("\n" + "=" * 60)
    print("BƯỚC 2: Parse văn bản -> Legal chunks + semantic sub-chunks")
    print("=" * 60)

    # Lưu toàn bộ chunk từ tất cả văn bản.
    all_chunks = []

    # Ánh xạ mỗi văn bản với danh sách chunk thuộc văn bản đó.
    # Map này sẽ được sử dụng khi tạo entity và relationship.
    doc_to_chunks_map = {}

    # Đếm số văn bản parse thành công.
    docs_with_chunks = 0

    # Đếm số văn bản không tạo được chunk.
    docs_without_chunks = 0

    # Duyệt qua từng văn bản trong DataFrame.
    for _, row in tqdm(
        df_content.iterrows(),
        total=len(df_content),
        desc="Đang parse văn bản"
    ):
        # Lấy số ký hiệu của văn bản.
        doc_id = row.get("Số Ký Hiệu")

        # Bỏ qua nếu số ký hiệu bị thiếu.
        if pd.isna(doc_id):
            docs_without_chunks += 1
            continue

        # Chuẩn hóa số ký hiệu thành chuỗi.
        doc_id = str(doc_id).strip()

        # Bỏ qua số ký hiệu rỗng hoặc chuỗi "nan".
        if not doc_id or doc_id.lower() == "nan":
            docs_without_chunks += 1
            continue

        # Parse nội dung văn bản thành danh sách chunk.
        chunks = parse_to_chunks(
            content=row.get("Content"),
            so_ky_hieu=doc_id
        )

        if chunks:
            # Thêm chunk của văn bản hiện tại vào danh sách tổng.
            all_chunks.extend(chunks)

            # Lưu ánh xạ giữa số hiệu văn bản và source_id của các chunk.
            doc_to_chunks_map[doc_id] = [
                chunk["source_id"]
                for chunk in chunks
            ]

            docs_with_chunks += 1

        else:
            # Văn bản không có nội dung hoặc không nhận diện được Điều.
            docs_without_chunks += 1

    # In thống kê kết quả.
    print(
        f"  → Tạo {len(all_chunks)} chunks "
        f"từ {docs_with_chunks} văn bản"
    )

    print(
        f"  → {docs_without_chunks} văn bản không tạo được chunk "
        f"(nội dung rỗng hoặc không nhận diện được Điều)"
    )

    return all_chunks, doc_to_chunks_map
```

```python
import asyncio
import json
import os
import random
from lightrag import LightRAG, QueryParam
from lightrag.llm.ollama import ollama_model_complete
from lightrag.operate import merge_nodes_and_edges
from dataclasses import asdict
from lightrag.utils import EmbeddingFunc
import numpy as np

# --- CẤU HÌNH HUGGINGFACE EMBEDDING ---
from langchain_community.embeddings import HuggingFaceEmbeddings

print("Đang khởi tạo mô hình HuggingFace Embedding...")
hf_model = HuggingFaceEmbeddings(
    model_name="MinhNguyet3001/embedding-supper-legal",
    # model_kwargs={"device": "cuda"}, # Dùng GPU để embedding nhanh hơn
    encode_kwargs={"normalize_embeddings": True}
)

# --- CẤU HÌNH CLOUD DATABASES (CHO ĐỒ ÁN) ---
os.environ["MILVUS_URI"] = "https://in03-8565fe07b73ad37.serverless.aws-eu-central-1.cloud.zilliz.com"
os.environ["MILVUS_TOKEN"] = "c82eaacfb7d554682daf1a93cc89932445482d69a8e66f44bf1dd4cb2c10d20b3ca29b1b468cd3b475c9fb92c45457cb485f97c0"
os.environ["MILVUS_DB_NAME"] = ""

os.environ["NEO4J_URI"] = "neo4j+s://698640cd.databases.neo4j.io"
os.environ["NEO4J_USERNAME"] = "698640cd"
os.environ["NEO4J_PASSWORD"] = "NlfgMfDoTWaz9-ITUUAwnNnNWlQB7RK8HMhrYLYXNKk"

# --- GIỚI HẠN CONCURRENCY CHUNG (AN TOÀN CHO EMBEDDING) ---
embedding_semaphore = asyncio.Semaphore(4)


async def custom_hf_embedding(
    texts: list[str],
) -> np.ndarray:
    if not texts:
        return np.empty(
            (0, 768),
            dtype=np.float32,
        )

    async with embedding_semaphore:
        # embed_documents là hàm đồng bộ.
        # Chạy trong thread để không khóa event loop.
        results = await asyncio.to_thread(
            hf_model.embed_documents,
            texts,
        )

    return np.asarray(
        results,
        dtype=np.float32,
    )


# --- CƠ CHẾ XOAY VÒNG API KEY ---
API_KEYS = [
    "91de644a6df04c049d14b3f1cd5e1350.PKXiI0wiDE2xvnBIIcJTh7eJ", #decomchay,
    "146aaac9913942bf9cca91292703ea7f.PDhYOFtqZVfAE9jiDOeJqRMc", #Nhi
    "304f63931e684534860521fe9d25d84f.8fI9g8QaR4cdc0vJ08nv0hqs", # Thiện 1
    "00c56b3effb049ff9df7bcd80b252f25.SjhN2hEoL70DfIMIrRhNLRQ4", #Quyen 1
    "bbd7b87643b0456e8371ee975332f0ed.szzaw4mCupQkBbRDMf06ofnm", #Quyen 2
    "9cdb97ea9b844e68aebc1b324726ca28.g3IqNlX3KS5DD0cNgU8yH-23", # Thien khôi
    "88394b02eb214ea2b0f2ed60303f9bd1.LPXrQMhhA_7w3ZB6zw9VShvX", #decomcom
    "832078448bb743bb93949efa1ed7e5f8.iTCprFu2PXvsHnZIAAhOskXE", # TÂN
    "87d92243078048cf9f5afa3ea859da3d.ILeaJqjRjnAqJ6EhnNf3iooX", #Mèn
    "d1e8ed57ead545f99e01ce4d984240f1.cvD032BSDi9QKWDUmIXsU9EL", #nguyetou
    "d2715dcab53e48038e1a988957d06095.v0cMgHXW_LMa-9biyqW4lj2y", #decomkawai
    "daa9c4d85a2646ba8095ee5dd292313a.u2h5E-HTtYfgIMV93pY6LZkv"  #decommcao
]
if not API_KEYS:
    raise ValueError(
        "Danh sách API_KEYS đang trống."
    )
 
current_key_idx = 0
key_rotation_lock = asyncio.Lock()
 
 
async def get_next_api_key() -> str:
    global current_key_idx
 
    async with key_rotation_lock:
        api_key = API_KEYS[current_key_idx]
 
        current_key_idx = (
            current_key_idx + 1
        ) % len(API_KEYS)
 
        return api_key
 
 
def is_rate_limit_error(error: Exception) -> bool:
    error_str = str(error).lower()
 
    keywords = [
        "429",
        "quota",
        "rate limit",
        "too many requests",
        "resource exhausted",
    ]
 
    return any(
        keyword in error_str
        for keyword in keywords
    )
 
 
def is_temporary_error(error: Exception) -> bool:
    error_str = str(error).lower()
 
    keywords = [
        "timeout",
        "timed out",
        "connection",
        "temporarily unavailable",
        "unavailable",
        "reset by peer",
        "502",
        "503",
        "504",
    ]
 
    return any(
        keyword in error_str
        for keyword in keywords
    )
 
 
async def rotating_ollama_model_complete(
    prompt,
    system_prompt=None,
    history_messages=None,
    **kwargs,
):
    if history_messages is None:
        history_messages = []
 
    max_retries = len(API_KEYS) * 2
    last_error = None
 
    for attempt in range(max_retries):
        api_key = await get_next_api_key()
 
        try:
            request_kwargs = dict(kwargs)
            request_kwargs["api_key"] = api_key
 
            result = await ollama_model_complete(
                prompt,
                system_prompt=system_prompt,
                history_messages=history_messages,
                **request_kwargs,
            )
 
            await asyncio.sleep(0)
 
            return result
 
        except Exception as error:
            last_error = error
 
            if is_rate_limit_error(error):
                wait_seconds = min(
                    1.5 * (attempt + 1),
                    10,
                )
 
                print(
                    f"⚠️ Key {api_key[:10]}... "
                    f"bị giới hạn. Chờ "
                    f"{wait_seconds:.1f}s rồi đổi key."
                )
 
                await asyncio.sleep(wait_seconds)
                continue
 
            if is_temporary_error(error):
                wait_seconds = min(
                    2 * (attempt + 1),
                    15,
                )
 
                print(
                    f"⚠️ Lỗi tạm thời: {error}. "
                    f"Thử lại sau {wait_seconds}s."
                )
 
                await asyncio.sleep(wait_seconds)
                continue
 
            raise
 
    raise RuntimeError(
        "Không gọi được LLM sau khi đã thử "
        f"toàn bộ API key. Lỗi cuối: {last_error}"
    )
 
 
# --- RETRY WRAPPER DÙNG CHUNG CHO CÁC THAO TÁC GHI DB (NEO4J/MILVUS) ---
async def retry_db_op(coro_func, *args, max_retries=5, label="db_op", **kwargs):
    """
    Chạy 1 coroutine ghi DB (Neo4j/Milvus) với retry + backoff cho lỗi
    mạng/timeout tạm thời. Lỗi không phải tạm thời sẽ raise ngay.
    """
    last_error = None
 
    for attempt in range(max_retries):
        try:
            return await coro_func(*args, **kwargs)
        except Exception as error:
            last_error = error
 
            if is_temporary_error(error) or is_rate_limit_error(error):
                wait_seconds = min(2 * (attempt + 1), 20) + random.uniform(0, 1)
                print(
                    f"⚠️ [{label}] lỗi tạm thời (lần {attempt + 1}/{max_retries}): "
                    f"{error}. Chờ {wait_seconds:.1f}s rồi thử lại."
                )
                await asyncio.sleep(wait_seconds)
                continue
 
            # Lỗi không rõ nguồn gốc -> vẫn thử lại vài lần cho an toàn,
            # nhưng log rõ để dễ debug nếu nó lặp lại.
            print(f"⚠️ [{label}] lỗi không xác định: {error}")
            wait_seconds = min(2 * (attempt + 1), 20)
            await asyncio.sleep(wait_seconds)
 
    raise RuntimeError(
        f"[{label}] Thất bại sau {max_retries} lần thử. Lỗi cuối: {last_error}"
    )
 
 
# --- GIỚI HẠN CONCURRENCY KHI GHI VÀO NEO4J/MILVUS ---
# Neo4j Aura free/serverless và Milvus serverless đều có giới hạn connection/
# request đồng thời. Chạy 30 task cùng lúc (batch_size cũ) dễ gây timeout
# hàng loạt. Giới hạn xuống 1 con số an toàn hơn nhưng vẫn song song.
graph_write_semaphore = asyncio.Semaphore(8)
 
 
async def run_in_batches(
    items,
    worker,
    batch_size=30,
):
    """
    Chạy `worker` cho từng item, giới hạn concurrency thực sự bằng
    graph_write_semaphore (không chỉ chia theo batch_size), để không làm
    quá tải Neo4j/Milvus dù batch_size lớn.
    """
    items = list(items)
 
    async def guarded_worker(item):
        async with graph_write_semaphore:
            await worker(item)
 
    for start in range(0, len(items), batch_size):
        current_batch = items[start:start + batch_size]
 
        # return_exceptions=True: 1 item lỗi (sau khi đã retry ở worker)
        # không được phép làm sập cả batch/pipeline.
        results = await asyncio.gather(
            *[guarded_worker(item) for item in current_batch],
            return_exceptions=True,
        )
 
        errors = [r for r in results if isinstance(r, Exception)]
        if errors:
            print(
                f"⚠️ {len(errors)}/{len(current_batch)} item trong batch "
                f"[{start}:{start + batch_size}] thất bại sau retry. "
                f"Ví dụ lỗi: {errors[0]}"
            )
 
 
async def build_macro_layer(
    rag,
    custom_entities_dict,
    custom_relations_dict,
):
    print(
        "\n--- GIAI ĐOẠN 2: "
        "NẠP TẦNG VĨ MÔ (MACRO) LÊN CLOUD ---"
    )
 
    print(
        "1. Nạp Entities/Relations Vĩ Mô "
        "vào Milvus VectorDB..."
    )
 
    vector_tasks = []
 
    if custom_entities_dict:
        vector_tasks.append(
            retry_db_op(
                rag.entities_vdb.upsert,
                custom_entities_dict,
                label="milvus_entities_upsert",
            )
        )
 
    if custom_relations_dict:
        vector_tasks.append(
            retry_db_op(
                rag.relationships_vdb.upsert,
                custom_relations_dict,
                label="milvus_relations_upsert",
            )
        )
 
    if vector_tasks:
        await asyncio.gather(*vector_tasks)
 
    print(
        "2. Nạp Node/Edge Vĩ mô "
        "vào Neo4j GraphDB..."
    )
 
    async def upsert_node(ent):
        await retry_db_op(
            rag.chunk_entity_relation_graph.upsert_node,
            ent["entity_name"],
            ent,
            label=f"neo4j_upsert_node:{ent.get('entity_name', '?')}",
        )
 
    async def upsert_edge(rel):
        await retry_db_op(
            rag.chunk_entity_relation_graph.upsert_edge,
            rel["src_id"],
            rel["tgt_id"],
            rel,
            label=f"neo4j_upsert_edge:{rel.get('src_id', '?')}->{rel.get('tgt_id', '?')}",
        )
 
    # Tạo node trước (giới hạn concurrency qua graph_write_semaphore).
    await run_in_batches(
        custom_entities_dict.values(),
        upsert_node,
        batch_size=30,
    )
 
    # Sau đó mới tạo edge.
    await run_in_batches(
        custom_relations_dict.values(),
        upsert_edge,
        batch_size=30,
    )
 
    print("-> Hoàn tất nạp tầng vĩ mô.")
 
 
CHECKPOINT_FILE = os.path.join(
    os.getcwd(),
    "micro_checkpoint.json",
)
FAILED_BATCHES_FILE = os.path.join(
    os.getcwd(),
    "micro_failed_batches.json",
)
 
 
def load_completed_batches() -> set[int]:
    if not os.path.exists(CHECKPOINT_FILE):
        return set(range(24))
 
    try:
        with open(CHECKPOINT_FILE, "r", encoding="utf-8") as file:
            data = json.load(file)
 
        return {
            int(idx)
            for idx in data.get("completed_batches", [])
        }
 
    except Exception as error:
        print(f"⚠️ Không đọc được checkpoint: {error}")
        return set(range(24))
 
 
def save_completed_batches(completed_batches: set[int]):
    temporary_file = CHECKPOINT_FILE + ".tmp"
 
    data = {
        "completed_batches": sorted(completed_batches),
        "completed_display": [
            idx + 1 for idx in sorted(completed_batches)
        ],
    }
 
    with open(temporary_file, "w", encoding="utf-8") as file:
        json.dump(data, file, ensure_ascii=False, indent=2)
 
    os.replace(temporary_file, CHECKPOINT_FILE)
 
 
def save_failed_batch(idx: int, error: str):
    """Ghi lại lô thất bại (sau khi đã retry) để có thể xem/chạy lại riêng."""
    failed = {}
 
    if os.path.exists(FAILED_BATCHES_FILE):
        try:
            with open(FAILED_BATCHES_FILE, "r", encoding="utf-8") as f:
                failed = json.load(f)
        except Exception:
            failed = {}
 
    failed[str(idx)] = str(error)
 
    temporary_file = FAILED_BATCHES_FILE + ".tmp"
    with open(temporary_file, "w", encoding="utf-8") as f:
        json.dump(failed, f, ensure_ascii=False, indent=2)
 
    os.replace(temporary_file, FAILED_BATCHES_FILE)
 
 
extraction_semaphore = asyncio.Semaphore(2)
merge_lock = asyncio.Lock()
checkpoint_lock = asyncio.Lock()
 
 
async def process_micro_batch(
    rag,
    batch_dict,
    idx,
    total_batches,
    json_path,
    completed_batches,
    max_batch_retries=3,
):
    """
    Xử lý 1 lô: extraction (song song, giới hạn bởi extraction_semaphore)
    rồi merge (tuần tự, giới hạn bởi merge_lock).
    Toàn bộ được bọc try/except + retry riêng cho từng lô: nếu 1 lô lỗi
    hẳn sau nhiều lần thử, nó KHÔNG làm crash các lô khác đang chạy song
    song — chỉ được ghi vào file lỗi để xử lý lại sau.
    """
    last_error = None
 
    for attempt in range(max_batch_retries):
        try:
            async with extraction_semaphore:
                print(
                    f"\n🚀 Extraction lô "
                    f"{idx + 1}/{total_batches} "
                    f"(lần thử {attempt + 1}/{max_batch_retries})"
                )
 
                pipeline_status = {
                    "history_messages": [],
                    "request_pending": False,
                    "latest_message": "",
                }
 
                pipeline_status_lock = asyncio.Lock()
 
                chunk_results = await rag._process_extract_entities(
                    batch_dict,
                    pipeline_status=pipeline_status,
                    pipeline_status_lock=pipeline_status_lock,
                )
 
            async with merge_lock:
                print(f"⏳ Merge lô {idx + 1}/{total_batches}")
 
                await merge_nodes_and_edges(
                    chunk_results=chunk_results,
                    knowledge_graph_inst=rag.chunk_entity_relation_graph,
                    entity_vdb=rag.entities_vdb,
                    relationships_vdb=rag.relationships_vdb,
                    global_config=rag._build_global_config(),
                    full_entities_storage=rag.full_entities,
                    full_relations_storage=rag.full_relations,
                    doc_id=f"custom_import_batch_{idx}",
                    pipeline_status=pipeline_status,
                    pipeline_status_lock=pipeline_status_lock,
                    llm_response_cache=rag.llm_response_cache,
                    entity_chunks_storage=rag.entity_chunks,
                    relation_chunks_storage=rag.relation_chunks,
                    current_file_number=1,
                    total_files=1,
                    file_path=json_path,
                )
 
                # Flush entities/relations local xuống đĩa ngay trong lúc
                # vẫn giữ merge_lock, để 2 lô chạy song song không cùng
                # ghi đè file JSON cùng lúc (race condition).
                await retry_db_op(
                    rag.full_entities.index_done_callback,
                    label=f"flush_full_entities_batch_{idx}",
                )
                await retry_db_op(
                    rag.full_relations.index_done_callback,
                    label=f"flush_full_relations_batch_{idx}",
                )
 
            async with checkpoint_lock:
                completed_batches.add(idx)
                save_completed_batches(completed_batches)
 
            print(f"✅ Hoàn tất lô {idx + 1}")
            return
 
        except Exception as error:
            last_error = error
            wait_seconds = min(3 * (attempt + 1), 30)
            print(
                f"❌ Lô {idx + 1} lỗi (lần thử {attempt + 1}/"
                f"{max_batch_retries}): {error}. "
                f"Chờ {wait_seconds}s rồi thử lại lô này."
            )
            await asyncio.sleep(wait_seconds)
 
    # Hết số lần retry cho lô này -> ghi nhận thất bại, KHÔNG raise,
    # để các lô khác vẫn tiếp tục chạy bình thường.
    print(
        f"🛑 Lô {idx + 1} thất bại sau {max_batch_retries} lần thử. "
        f"Bỏ qua, sẽ cần chạy lại lô này sau. Lỗi cuối: {last_error}"
    )
    save_failed_batch(idx, str(last_error))
 
 
async def build_micro_layer(
    rag,
    custom_chunks_dict,
    json_path,
    reload_base_storage=True,
):
    print(
        "\n--- GIAI ĐOẠN 3: "
        "NẠP TẦNG VI MÔ (MICRO) LÊN CLOUD ---"
    )
 
    if reload_base_storage:
        print("3. Nạp Data Nguyên Bản vào Local Text Storage...")
 
        if custom_chunks_dict:
            await retry_db_op(
                rag.text_chunks.upsert,
                custom_chunks_dict,
                label="text_chunks_upsert",
            )
            # Flush ngay xuống kv_store_text_chunks.json trên đĩa, để dù
            # có dừng script ở bước sau (extraction) thì file này vẫn còn.
            await retry_db_op(
                rag.text_chunks.index_done_callback,
                label="text_chunks_flush",
            )
 
        print("4. Nạp Embeddings Data Nguyên Bản vào Milvus VectorDB...")
 
        if custom_chunks_dict:
            await retry_db_op(
                rag.chunks_vdb.upsert,
                custom_chunks_dict,
                label="chunks_vdb_upsert",
            )
 
    else:
        print("⏭️ Bỏ qua upsert text/chunk vectors vì đang chạy resume.")
 
    print("5. Kích hoạt LLM Extraction...")
 
    if not custom_chunks_dict:
        print("⚠️ Không có chunk để xử lý.")
        return
 
    BATCH_SIZE = 20
 
    items = list(custom_chunks_dict.items())
 
    batches = [
        dict(items[i:i + BATCH_SIZE])
        for i in range(0, len(items), BATCH_SIZE)
    ]
 
    completed_batches = load_completed_batches()
 
    pending_batches = [
        (idx, batch_dict)
        for idx, batch_dict in enumerate(batches)
        if idx not in completed_batches
    ]
 
    print(f"Tổng chunks: {len(items)}")
    print(f"Tổng lô: {len(batches)}")
    print(f"Đã hoàn tất: {len(completed_batches)} lô")
    print(f"Còn lại: {len(pending_batches)} lô")
 
    if not pending_batches:
        print("✅ Không còn lô nào cần xử lý.")
        return
 
    tasks = [
        asyncio.create_task(
            process_micro_batch(
                rag=rag,
                batch_dict=batch_dict,
                idx=idx,
                total_batches=len(batches),
                json_path=json_path,
                completed_batches=completed_batches,
            )
        )
        for idx, batch_dict in pending_batches
    ]
 
    # return_exceptions=True: process_micro_batch tự bắt lỗi nội bộ rồi,
    # nhưng để chắc chắn tuyệt đối không có gì thoát ra làm sập gather.
    await asyncio.gather(*tasks, return_exceptions=True)
 
    final_completed = load_completed_batches()
    remaining = [idx for idx, _ in pending_batches if idx not in final_completed]
 
    if remaining:
        print(
            f"\n⚠️ Còn {len(remaining)} lô chưa hoàn tất "
            f"(xem chi tiết trong {FAILED_BATCHES_FILE}). "
            f"Chạy lại script để tự động resume các lô này."
        )
    else:
        print("\n-> Hoàn tất rút trích & nạp tầng vi mô!")
 
 
async def main():
    json_path = os.path.join(os.getcwd(), "custom_kg_full.json")
    if not os.path.exists(json_path):
        print(f"Lỗi: Không tìm thấy file {json_path}. Vui lòng chạy các block phía trên trước.")
        return
 
    print(f"Đang tải dữ liệu từ {json_path}...")
    with open(json_path, "r", encoding="utf-8") as f:
        custom_kg = json.load(f)
 
    custom_entities_dict = {
        ent["entity_name"]: {
            **ent,
            "entity_id": ent["entity_name"],
            "content": f"{ent['entity_name']}: {ent.get('description', '')}",
            "file_path": ent.get("file_path", "unknown_source"),
        }
        for ent in custom_kg.get("entities", [])
    }
 
    custom_relations_dict = {
        f"{rel['src_id']}-{rel['tgt_id']}": {
            **rel,
            "content": rel.get("description", ""),
            "file_path": rel.get("file_path", "unknown_source"),
        }
        for rel in custom_kg.get("relationships", [])
    }
 
    custom_chunks_dict = {
        chunk["source_id"]: {
            "content": chunk.get("content", ""),
            "source_id": chunk["source_id"],
            "file_path": chunk.get("file_path", "unknown_source"),
            "full_doc_id": chunk.get("file_path", chunk["source_id"]),
            "chunk_order_index": chunk.get("chunk_order_index", 0),
        }
        for chunk in custom_kg.get("chunks", [])
        if chunk.get("source_id")
    }
 
    WORKING_DIR = "./target_db"
    if not os.path.exists(WORKING_DIR):
        os.mkdir(WORKING_DIR)
 
    rag = LightRAG(
        working_dir=WORKING_DIR,
        llm_model_max_async=3,
        vector_storage="MilvusVectorDBStorage",
        graph_storage="Neo4JStorage",
        llm_model_func=rotating_ollama_model_complete,
        llm_model_name="gpt-oss:120b-cloud",
        embedding_func=EmbeddingFunc(
            embedding_dim=768,
            max_token_size=256,
            func=custom_hf_embedding,
        ),
    )
 
    print("\nĐang khởi tạo các kết nối tới Milvus, Neo4j...")
    await rag.initialize_storages()
 
    print("\nBắt đầu Pipeline Data Ingestion...")
    await build_macro_layer(rag, custom_entities_dict, custom_relations_dict)
    await build_micro_layer(rag, custom_chunks_dict, json_path)
 
    print("\nĐang flush toàn bộ dữ liệu local xuống đĩa (kv_store_*.json)...")
    for storage_name, storage_obj in [
        ("text_chunks", rag.text_chunks),
        ("full_entities", rag.full_entities),
        ("full_relations", rag.full_relations),
        ("llm_response_cache", rag.llm_response_cache),
    ]:
        try:
            await retry_db_op(
                storage_obj.index_done_callback,
                label=f"final_flush_{storage_name}",
            )
        except Exception as error:
            print(f"⚠️ Không flush được {storage_name}: {error}")
 
    print("\n🚀 HOÀN TẤT NHÚNG DỮ LIỆU THÀNH CÔNG LÊN CLOUD (MILVUS & NEO4J)!")
 
```
